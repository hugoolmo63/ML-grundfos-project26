```python
# Defining main Data Frames

test1 = pd.read_csv(test1_p)
train1 = pd.read_csv(train1_p)

test2 = pd.read_csv(test2_p)
train2 = pd.read_csv(train2_p)

# print('Test 1:', test1.head())
# print('Train 1:', train1.head())

# print('Test 2:', test2.head())
# print('Train 2:', train2.head())

# print(train1.describe())
# print(train2.describe())
```

```python
# filtering and cleaning

raw_train1, raw_train2 = train1, train2

test1 = test1[test1['operating_state'] == 'stable'].replace([float('inf'), -float('inf')], float('nan')).dropna().drop_duplicates().reset_index(drop=True)
train1 = train1[train1['operating_state'] == 'stable'].replace([float('inf'), -float('inf')], float('nan')).dropna().drop_duplicates().reset_index(drop=True)
test2 = test2[test2['operating_state'].isin(['stable','transient'])].replace([float('inf'), -float('inf')], float('nan')).dropna().drop_duplicates().reset_index(drop=True)
train2 = train2[train2['operating_state'].isin(['stable','transient'])].replace([float('inf'), -float('inf')], float('nan')).dropna().drop_duplicates().reset_index(drop=True)
```

## Task 0: Recovering pump physics

The target is pump power `P`. Speed is converted from rpm to krpm before constructing the polynomial features required by the assignment.

```python
from itertools import combinations

from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

task0_train = train1.copy()
task0_test = test1.copy()

for frame in (task0_train, task0_test):
    frame['n_krpm'] = frame['speed_rpm'] / 1000

polynomial_terms = [(i, j) for i in range(4) for j in range(4)]

def polynomial_design(frame):
    n = frame['n_krpm'].to_numpy()
    q = frame['flow_m3_h'].to_numpy()
    return np.column_stack([n ** i * q ** j for i, j in polynomial_terms])

X_train = polynomial_design(task0_train)
X_test = polynomial_design(task0_test)
y_train = task0_train['power_w'].to_numpy()
y_test = task0_test['power_w'].to_numpy()

def regression_metrics(y_true, prediction):
    return {
        'MAE (W)': mean_absolute_error(y_true, prediction),
        'RMSE (W)': mean_squared_error(y_true, prediction) ** 0.5,
        'R2': r2_score(y_true, prediction),
    }

display(pd.DataFrame({
    'term': [f'n^{i} Q^{j}' for i, j in polynomial_terms],
    'feature': [f'n_{i}q_{j}' for i, j in polynomial_terms],
}))
```

### Linear baselines and BIC feature selection

The baseline uses the original linear variables `n` and `Q`. For the polynomial model, every subset of the 16 allowed terms is fitted on the training data and ranked with BIC. This penalizes unnecessary terms while keeping the selection entirely within the training set.

```python
baseline = LinearRegression().fit(task0_train[['n_krpm', 'flow_m3_h']], y_train)
baseline_prediction = baseline.predict(task0_test[['n_krpm', 'flow_m3_h']])

n_train = len(y_train)
selection_results = []
for subset_size in range(1, len(polynomial_terms) + 1):
    for selected in combinations(range(len(polynomial_terms)), subset_size):
        selected = np.array(selected)
        coefficients = np.linalg.lstsq(X_train[:, selected], y_train, rcond=None)[0]
        residuals = y_train - X_train[:, selected] @ coefficients
        rss = np.sum(residuals ** 2)
        bic = n_train * np.log(rss / n_train) + subset_size * np.log(n_train)
        selection_results.append((bic, selected, coefficients))

best_bic, selected_indices, selected_coefficients = min(selection_results, key=lambda result: result[0])
selected_train = X_train[:, selected_indices]
selected_test = X_test[:, selected_indices]
selected_model = LinearRegression(fit_intercept=False).fit(selected_train, y_train)
selected_prediction = selected_model.predict(selected_test)

full_model = LinearRegression(fit_intercept=False).fit(X_train, y_train)
full_prediction = full_model.predict(X_test)

metrics = pd.DataFrame({
    'linear baseline': regression_metrics(y_test, baseline_prediction),
    'full 16-term polynomial': regression_metrics(y_test, full_prediction),
    'BIC-selected polynomial': regression_metrics(y_test, selected_prediction),
}).T
display(metrics)
print('Selected terms:', [polynomial_terms[i] for i in selected_indices])
print('BIC:', best_bic)
```

### Final regression expression

The coefficients below use `n` in krpm and `Q` in m³/h, so the expression can be reported directly in the requested form.

```python
def format_term(coefficient, i, j):
    sign = '+' if coefficient >= 0 else '-'
    magnitude = abs(coefficient)
    factors = []
    if i:
        factors.append('n' if i == 1 else f'n^{i}')
    if j:
        factors.append('Q' if j == 1 else f'Q^{j}')
    return f' {sign} {magnitude:.8g}' + (f'*' + '*'.join(factors) if factors else '')

expression = ''.join(
    format_term(coefficient, *polynomial_terms[index])
    for index, coefficient in zip(selected_indices, selected_model.coef_)
)
print(f'P(n, Q) = {expression.lstrip(" +")}')
print('where n is in krpm, Q is in m^3/h, and P is in W')
```
